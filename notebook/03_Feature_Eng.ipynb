```python
data_frame['AppointmentDateTime'] =pd.to_datetime(data_frame['AppointmentDateTime'])
```

```python
data_frame['AppointmentHour'] = data_frame['AppointmentDateTime'].dt.hour
data_frame['AppointmentDayOfWeek'] = data_frame['AppointmentDateTime'].dt.day_name()
data_frame['AppointmentMonth'] = data_frame['AppointmentDateTime'].dt.month_name()
```

```python
data_frame.head()
```

```python
data_frame['IsWeekend'] = data_frame['AppointmentDayOfWeek'].isin(['Saturday', 'Sunday'])
```

```python
data_frame.head()
```

```python
data_frame.head()
```

```python
def lead_time_booking(days):
    if days == 0:
        return 'Same Day'
    elif days <= 7:
        return 'Within a Week'
    elif days <= 30:
        return 'Within a Month'
    else:
        return 'Over a Month'

data_frame['LeadTimeCategory'] = data_frame['BookingLeadTimeDays'].apply(lead_time_booking)
```

```python
data_frame['IsNoShow'] = (data_frame['Status'] == 'No-Show').astype(int)
```
